```python
import numpy as np
import sympy as sp
from IPython.display import Math, Markdown # Affichage de formules mathématiques amelioré
```

```python
xx = np.array(sorted([1, 3, 4, 2, 2, 4, 5, 5, 1, 1, 2, 3 ,4, 3, 2, 6 ,6, 1, 2, 2, 3,2,1,3,4,2,3,3,3,5,6,4,2,2,4,2]))

n = len(xx)
display(xx)
display(n)
```

```python
aa, nn = np.unique(xx, return_counts=True)

display(Markdown(f"Les valeurs uniques sont {aa} et leur effectif est {nn}"))
```

```python
# Moyenne arithmétique
mu = sum(xx)/sp.S(n) # sp.S : dire que c'est un symbole
# mu = sp.Rational(sum(xx), n)
display(Markdown(f"La moyenne est {mu} "))
```

```python
mode = aa[np.argmax(nn)]
print(f'Le mode est {mode}')
```

```python

mediane = np.median(xx)
print(f'La médiane est {mediane}')
```

```python
# Variance

# Formule basique
variance_1 = sum([(x-mu) **2 for x in xx])/sp.S(n)
# Formule Huguyens
variance_2 = sum([x**2 for x in xx])/sp.S(n)-mu**2

if variance_1 != variance_2: display("ERROR")

variance_np = np.var(xx)

display(Markdown(f"La variance avec sympy est ${sp.latex(variance_1)} \\approx {variance_1.evalf()}$, avec numpy {variance_np}"))
```

```python
# Ecart-type
s_sp = sp.sqrt(variance_1)
s_np = np.std(xx)
display(Markdown(f"L'écart-type avec sympy est ${sp.latex(s_sp)} \\approx{s_sp.evalf()}$ avec numpy {s_np}"))
```
