# F(x)=0, cas de la dimension 1

## Exercice 2

- Application de la méthode du point fixe

```python
F = lambda x : x**3-2
```

```python
omega=-1/5
print(omega)
```

```python
G = lambda t : t + omega * F(t)
```

```python
from time import sleep
```

```python
x=2
c=0
while abs(F(x)) > 1e-6 :
    x = G(x)
    print(x)
    sleep(.5)
    c+=1
print(x, c)
```
