# TP Tranformée de Fourier

```python
 

N=10
L=10
dx=L/N

x = linspace(dx/2,L-dx/2,N)
xFin = linspace(0, L, 1000)

U = []
for i in range(N) :
	if i<N/2 :
		U.append(0)
	else :
		U.append(1)

F = dct(U)
#print(F)

#Assemblage de la fonction
fct = F[0]/(2*N)
for i in range(1, N) :
	fct += F[i]/N * cos(pi*i*xFin/L)

print('Fourier inverse de fourier=identité',max(abs(0.5/N*idct(dct(U))-U)))
plot(x, F, '.');
plt.show()

print("Fonction d'interpolation")
plot(xFin, fct)
plot(x, U, 'o')
plt.show()
```
